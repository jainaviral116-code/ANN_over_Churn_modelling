```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('Churn_Modelling.csv')
```

```python
## EDA
X = dataset.iloc[:, 3:13]
Y = dataset.iloc[:, 13]
```

```python
X
```

```python
Y
```

```python
# handle categorial data with one-hot encoding
gender = pd.get_dummies(X['Gender'], dtype=int, drop_first=True)
geography = pd.get_dummies(X['Geography'], dtype = int, drop_first=True)
```

```python
geography
```

```python
X = X.drop(['Gender', 'Geography'], axis=1)
```

```python
X = pd.concat([X, gender, geography], axis=1)
```

```python
X.head()
```

```python
X.shape
```

```python
# Splitting data
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_train.shape
```

```python
# model training
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import ReLU, PReLU, LeakyReLU , ELU
from tensorflow.keras.layers import Dropout
```

```python
classifier = Sequential()

```

```python
classifier.add(Dense(units = 11, activation = 'relu'))


```

```python
classifier.add(Dense(units = 7, activation = 'relu'))
classifier.add(Dropout(0.2))
```

```python
classifier.add(Dense(units = 6, activation = 'relu'))
classifier.add(Dropout(0.3))
```

```python
classifier.add(Dense(units = 1, activation = 'sigmoid'))
```

```python
import tensorflow
opt = tensorflow.keras.optimizers.Adam(learning_rate = 0.01)
```

```python
classifier.compile(optimizer = opt, loss='binary_crossentropy', metrics=['accuracy'])
```

```python
import tensorflow as tf
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    min_delta=0.001,
    patience=20,
    verbose=1,
    mode="auto",
    baseline=None,
    restore_best_weights=False,
    start_from_epoch=0,
)
```

```python
model_history = classifier.fit(X_train, Y_train, validation_split = 0.33, batch_size=10, epochs = 1000, callbacks=early_stopping)
```

```python
model_history.history.keys()
```

```python
plt.plot(model_history.history['accuracy'])
plt.plot(model_history.history['val_accuracy'])
plt.title('model accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend(['train', 'test'], loc='upper left')
```

```python
plt.plot(model_history.history['loss'])
plt.plot(model_history.history['val_loss'])
plt.title('model loss')
plt.xlabel('epoch')
plt.ylabel('loss')
plt.legend(['train', 'test'], loc='upper left')
```

```python
pred = classifier.predict(X_test)
```

```python
pred
```

```python
y_pred = (pred >= 0.5)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(Y_test, y_pred)
```

```python
cm
```

```python
from sklearn.metrics import accuracy_score
score = accuracy_score(y_pred, Y_test)
```
